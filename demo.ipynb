حوّل أي نص عربي لصوت واضح: **صوت ست**، **بنت صغيرة**، أو **ولد صغير** — مجانًا بالكامل.

## طريقة التشغيل
1. من القائمة فوق اختار **Runtime → Run all** (أو دوس Ctrl+F9)
2. استنى دقيقتين لحد ما التثبيت وتحميل الموديل يخلصوا
3. في آخر خلية هيظهر رابط أزرق فيه `gradio.live` — افتحه واستخدم الواجهة

الرابط بيفضل شغال طول ما الصفحة دي مفتوحة (لحد ١٢ ساعة). لو قفلت، شغّل Run all من الأول.

```
# ٢) تحميل كود الواجهة والأصوات
import os, urllib.request

os.makedirs("voices", exist_ok=True)
GH = "https://raw.githubusercontent.com/bahyfree1/arabic-voice-studio/main"
HF = "https://bahyfree1-arabic-voice-studio.static.hf.space"

for f in ["web_app.py", "tashkeel_utils.py"]:
    urllib.request.urlretrieve(f"{GH}/{f}", f)
for name in ["female-ar", "child-girl-10", "child-boy-15"]:
    for ext in ["wav", "txt"]:
        urllib.request.urlretrieve(f"{HF}/voices/{name}.{ext}", f"voices/{name}.{ext}")
print("كل حاجة اتنزلت ✔")
```

```
# ٣) تشغيل الواجهة — الرابط العام هيظهر تحت (سطر فيه gradio.live)
!python web_app.py
```
